```python
def generators():
  yield 10
  yield 30
  yield 59
  yield 33

num=generators()
for i in num:
  print(i)
```

```python
def generators():
  yield 10
  yield 30
  yield 59
  yield "Hello"
  yield 33

x=generators()
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

```python
def gen():
  x=yield 10
  yield x
g=gen()

print(next(g))
print(g.send(59))
```
