```python
dic=dict({1:"Puja",2:"Ghanshyam",3:"Radha"})
print(dic)
print(dic.get(1))
print(dic.get(4))
print(dic[3])
```

```python
dic.update({4:"Sejal",5:"Tanisha"})
print(dic)
```

```python
my_dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
print(my_dic)
my_dic.pop("friend")
print(my_dic)
my_dic.popitem()
print(my_dic)
del my_dic["age"]
print(my_dic)
my_dic.clear()
print(my_dic)
```

```python
my_dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
print(my_dic.keys())
print(my_dic.values())
print(my_dic.items())
```

```python
dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
for i in dic:
  print(i, end=" ")
print()
for key in dic.keys():
  print(key,end=" ")
```

```python
dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
for i in dic.values():
  print(i)
```

```python
dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
for key,val in dic.items():
  print(key,val)
```

```python
dic={"name":"Puja","age":21,"friend":"Sejal" , 'course':"Java"}
dic["course"]="python"
print(dic)
new_dic=dic.copy()
new_dic['friend']='Saloni'
print(new_dic)
```

```python
nested_dic={
    "student1":{
        "name":"Lovely",
        "age":32,
        "city":"Bangalore"
    },
    "student2":{
        "name":"Puja",
        "age":33,
        "city":"Patna"
    }
}
print(nested_dic)
print(nested_dic["student1"])
print(nested_dic["student2"]["name"])
```

```python
dic={1:2,3:4,5:6}
new_dic={i:i**2 for i in dic}
print(new_dic)
```

```python
li=[2,3,4,5,6,7,8,9]
dic={i:i**2 for i in li}
print(dic)
```

```python
li=[2,3,4,5,6,7,8,9]
dic={i:i*i for i in li if i%2==0}
print(dic)
```

```python
dic={i:i**3 for i in range(1,21)}
print(dic)
```
