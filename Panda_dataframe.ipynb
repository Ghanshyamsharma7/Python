```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
    "name":["Puja","Saloni","Ragini","Niki","Gabbana","Lovely"],
    "city":["Patna","Delhi","Patna","Mumbai","Kolkata","Patna"],
    "age":[34,21,np.nan,23,np.nan,45],
    "salary": [50000, 62000, 58000, 71000, 120000, 62000],
    "dept":   ["IT", "HR", "IT", "Sales", "IT", "HR"],
})

print(df)
```

# Attribute

```python
print(df.shape)
print(df.columns)
print(df.size)
print(df.index)
print(df.ndim)
print(df.values)
print(len(df))
```

```python
print(df.T)
```
