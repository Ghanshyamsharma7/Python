```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
    "name":["Puja","Saloni","Ragini","Niki","Gabbana","Lovely"],
    "city":["Patna","Delhi","Patna","Mumbai","Kolkata","Patna"],
    "age":[34,21,np.nan,23,np.nan,45],
    "salary": [50000, 62000, 58000, 71000, 120000, 62000],
    "dept":   ["IT", "HR", "IT", "Sales", "IT", "HR"],
})

print(df)
```

# **filtering rows(boolean indexing)**

```python
print(df[df["salary"]>6000])
```

```python
print(df[(df["age"]>23) & (df['dept']=="IT")])
```

```python
print(df[(df["city"]=="Delhi") | (df['city']=="Patna")])
```

```python
print(df[~(df["dept"]=="HR")])
```

```python
df[df["city"].isin(["Delhi","Mumbai"])]
```

```python
df[df["salary"].between(50000,70000)]
```

```python
df[df["name"].str.startswith("P")]
```

```python
df[df['age'].isna()]
```

```python
df[df["age"].notna()]
```

```python
df[df["name"].str.contains("l")]
```

```python
df[df["name"].str.contains("l",case=False)]
```

```python
#alternate method for query
df.query("salary>60000 and dept=='IT'")
```

```python
threshold=60000
df.query("salary> @threshold")
```

```python
df.loc[df["salary"]>50000,["name","dept"]]
```
