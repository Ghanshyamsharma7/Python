```python
# exception handling in files
try:
  file=open("Java.txt")
  data=file.read()
  print(data)

except FileNotFoundError as e:
  print(e)
```

```python
with open("abc.txt","a") as f:
  f.write("This is appended line")

file=open("abc.txt")
contents=file.read()
print(contents)
file.close()
```

```python
with open("java.txt","a+") as f:
  f.write("This file has been created by append method")

file=open("java.txt")
data=file.read()
print(data)
```

```python
line="This file has been created by append method"
with open("java.txt","a+") as f:
  f.seek(0)
  if line not in f.read():
    f.write("line")
```

```python
with open("hindi.txt","x") as f:
  f.write("This is hindi file")

with open("hindi.txt") as f:
  data=f.read()
  print(data)
```

```python
try:
  with open("hindi.txt", "x") as f:
    f.write("This is not a hindi file")
except FileExistsError as e:
  print(e)
```

```python
line=["This is first line\n","This is second line\n",'this is third line\n']
with open("english.txt", "a") as f:
  f.writelines(line)

with open("english.txt") as f:
  print(f.read())
```
