```python
from abc import ABC, abstractmethod

class Greet(ABC):
  @abstractmethod
  def say(self):
    pass
class English(Greet):
  def say(self):
    return 'Hello Bro'

g=English()
g.say()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

  def move(self):           #concrete method
    return 'moving'

class Dog(Animal):
  def sound(self):
    return "barks"

d=Dog()
print(d.sound())
print(d.move())
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @property
  @abstractmethod
  def sound(self):
    pass
  @property
  def move(self):           #concrete method
    return 'moving'

class Dog(Animal):
  @property
  def sound(self):
    return "barks"

d=Dog()
print(d.sound)      # property makes the method as property we can call it without ()
print(d.move)
```
