```python
import numpy as np
a=np.array([23,33,44,55,77])
b=a[0:3]   #b is view of a (shares same memory)
b[0]=99
print(b)
print(a)
print(b.base)
print(b.base is a)
```

# **Copy and View**

```python
a=np.array([23,33,44,55,77])
b=a[0:3].copy()  # copy of a
b[0]=99
print(b)
print(a)
print(b.base)
print(b.base is a)
```

# **Slicing**

```python
arr=np.arange(20)
a=arr.reshape(5,4)
print(a[:])
print(a[::-1])   #reverse rows
print(a[1])      #second row
print(a[2,3])
print(a[:,2])    #third column
print(a[1:4,0:3])
```

```python
print(a[1:3])
print()
print(a[::2])
print()
print(a[::2,::2]) #every second row and column
```

```python
print(a)
print(a[::,::-1])  #Reverse column
```

```python
print(a[:,0])   #print first column
```

```python
print(a[...,0])  #ellipsis same as a[:,0]
```

# **Fancy indexing(Index with Arrays)**

```python
a=np.arange(20)
a=a.reshape(4,5)
print(a)
```

```python
print(a[[0,2,3]])
```

```python
print(a[[2,2,1,3]])      #repititon is allowed
```

```python
print(a[[0,1]])
```

```python
print(a[:,[0,2,3]])     #columns only
```

```python
print(a[[0,1],[1,2]])
```

```python
a=np.array([3,4,5,6,7])
a[[0,1]]=0
print(a)
```

```python
b=a[[2,3]]
b[0]=9
print(a)  #fancy indexing reading is copy
print(b)
```
