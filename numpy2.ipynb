# **Copy and View**

```python
a=np.array([23,33,44,55,77])
b=a[0:3].copy()  # copy of a
b[0]=99
print(b)
print(a)
print(b.base)
print(b.base is a)
```

# **Slicing**

```python
arr=np.arange(20)
a=arr.reshape(5,4)
print(a[:])
print(a[::-1])   #reverse rows
print(a[1])      #second row
print(a[2,3])
print(a[:,2])    #third column
print(a[1:4,0:3])
```

```python
print(a[1:3])
print()
print(a[::2])
print()
print(a[::2,::2]) #every second row and column
```

```python
print(a)
print(a[::,::-1])  #Reverse column
```

```python
print(a[:,0])   #print first column
```

```python
print(a[...,0])  #ellipsis same as a[:,0]
```

# **Boolean Masking**

```python
a=np.array([33,9,44,7,15,6])

mask=a>10
print(mask)
print(a[mask])
```

```python
print(a>10)
print(a[a>10])
```

```python
print(a[(a<70) & (a>10)])
print(a[(a<10) | (a>30)])
print(a[~(a>10)])
```

# **Modify values using mask**

```python
a=np.array([23,7,6,55,43,4,32,21])
a[a>10]=20            #capp the values at 20 (clipping)
print(a)
```

```python
a=np.array([11,23,44,54,3,21,7])
a[a<10]=0
print(a)
```

```python
a=np.array([11,23,44,54,3,21,7])
a[a%2==0]=0  #zero out all even numbers
print(a)
```

# **np.where()**

```python
a=np.array([11,23,44,54,3,21,7])
x=np.where(a>10)  #returns tuple of indexes
print(x)
```

```python
z=np.where(a>10,"Yes", "No")
print(z)
```

```python
z=np.where(a>20,a,0)
print(z)
```

```python
arr=np.array([3,0,4,5,0,0,1])
a=np.nonzero(arr)   #return the tuple of index
print(a)
```
