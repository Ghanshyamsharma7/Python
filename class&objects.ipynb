```python
class student:
  college="WCTM"                   #class variables

  def __init__(self,name,age):
    self.name=name                 #instance variable
    self.age=age

  def show(self):                  #method
    print("Name:", self.name, "Age:", self.age)

stu1=student("Puja",21)             #object
stu2=student("Ghanshyam",22)
stu1.show()
stu2.show()

stu1.college="IIT Bombay"

print(stu1.college)
print(stu2.college)
```

```python
class circle:
  def __init__(self,radius):
    self.r=radius

  def area(self):
    return 3.14*self.r*self.r

cir1=circle(5)
cir1.area()
```

```python
#__str__() method to print the object
class employee:
  def __init__(self,name,salary):
    self.n=name
    self.s=salary
  def __str__(self):
    return f"Name:{self.n} Salary:{self.s}"

emp=employee("Puja",100000)
print(emp)
```

```python
class student:
  college="WCTM"                   #class variables

  def __init__(self,name,age,salary=100000):
    self.name=name                 #instance variable
    self.age=age
    self.salary=salary

  def show(self):                  #method
    print("Name:", self.name, "Age:", self.age, "salary:",self.salary)

stu1=student("Puja",21,20000)             #object
stu2=student("Ghanshyam",22)
stu1.show()
stu2.show()

stu1.college="IIT Bombay"

print(stu1.college)
print(stu2.college)
```

```python
class user:
  def __init__(self,name):
    self.n=name

  def greet(self):
    return f"Hello {self.n}"

  def welcome(self):
    self.greet()
    print("Welcome to the Python class")

u=user("Puja")
u.age=32                      #updating instance properties
u.welcome()
print(u.age)
```

```python
class calc:
  def sum(self,a,b):
    return a+b

  def mul(self,a,b):
    return a*b

cal=calc()
print(cal.sum(3,4))
print(cal.mul(7,8))
```

```python
#class methods
class student:
  total=0

  @classmethod
  def final_mark(cls,mark):
    cls.total+=mark
    return cls.total

print(student.final_mark(80))
print(student.final_mark(90))
```

```python
# static methods
class cal:

  @staticmethod
  def fun(a,b):
    return (a*b,a/b,a+b,a-b)

obj=cal()
obj.fun(7,8)
cal.fun(8,2)
```
