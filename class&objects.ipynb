```python
class student:
  college="WCTM"                   #class variables

  def __init__(self,name,age):
    self.name=name                 #instance variable
    self.age=age

  def show(self):                  #method
    return self.name, self.age

stu=student("Puja",21)             #object
stu.show()

student.college="IIT Bombay"

print(stu.college)
```

```python
class student:
  college="WCTM"                   #class variables

  def __init__(self,name,age):
    self.name=name                 #instance variable
    self.age=age

  def show(self):                  #method
    print("Name:", self.name, "Age:", self.age)

stu1=student("Puja",21)             #object
stu2=student("Ghanshyam",22)
stu1.show()
stu2.show()

stu1.college="IIT Bombay"

print(stu1.college)
print(stu2.college)
```

```python
class circle:
  def __init__(self,radius):
    self.r=radius

  def area(self):
    return 3.14*self.r*self.r

cir1=circle(5)
cir1.area()
```

```python
class employee:
  def __init__(self,name,salary):
    self.n=name
    self.s=salary

emp=employee("Puja",100000)
print(emp)
```

```python
#__str__() method to print the object
class employee:
  def __init__(self,name,salary):
    self.n=name
    self.s=salary
  def __str__(self):
    return f"Name:{self.n} Salary:{self.s}"

emp=employee("Puja",100000)
print(emp)
```
