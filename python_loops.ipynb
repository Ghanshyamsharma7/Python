```python
#9.Find the factorial of a number entered by the user.
n=int(input("enter a number: "))
product=1
for i in range(1,n+1):
  product*=i
print("The factorial of ",n, "is" ,product)
```

```python
#10.Count the digits in a number (use while)
num=int(input("enter a number "))
count=0
if num==0:
  count=1
while num>0:
  count+=1
  num//=10
print(count)
```

```python
#11.Find the sum of the digits of a number
num=input('enter a number')
sum=0
for i in num:
  sum+=int(i)
print(sum)
```

```python
#12.Reverse a number

num=input("enter a num ")
for i in range(len(num)):
  num=num[::-1]
print(num)
```

```python
#13.Check whether a number is a palindrome
num=input("enter num ")
if num==num[::-1]:
   print("Pallindrome")
else:
    print("Not a Pallindrome")
```

```python
#14.Count the vowels and consonants in a string.
text=input("enter a string: ")
vowels="aeiouAEIOU"
v=c=0
for ch in text:
  if ch.isalpha():
    if ch in vowels:
      v+=1
    else:
      c+=1
print("vowels:",v)
print("consonants:",c)
```

```python
#16.Find the largest and smallest of 5 numbers entered by the user.

li=[]
for i in range(5):
  num=int(input('enter number '))
  li.append(num)
print(li)

min=li[0]
max=li[0]
for i in range(len(li)):
  if min>li[i]:
    min=li[i]
  if max<li[i]:
    max=li[i]
print("Smallest: ",min)
print('Largest: ',max)

```

```python
#18.Calculate the average of numbers entered by the user; stop when they type 0
n=10
li=[]
for i in range(n):
  num=int(input("enter non-zero numbers"))
  if num==0:
    break
  li.append(num)

sum=0
for i in li:
  sum+=i
  avg=sum/len(li)
print(avg)
```
