```python
#14.Number guessing game: the computer picks a random number from 1-100 (random.randint); the user keeps guessing and gets “too high” or “too low” hints, plus a count of attempts.
import random

num=random.randint(1,10)
count=0
while True:
  n=int(input("enter a number: "))
  count+=1
  if n>(num):
    print("too high")
  elif n<(num):
    print("too low")
  elif num==n:
    print("You guessed right! in attempt ", count)
    break
```

```python
#15.Simple ATM menu: use a while True loop with options Check Balance, Deposit, Withdraw, Exit.

balance=10000

while True:
  print("Welcome to SBI ATM")
  print("""
  1.Balance
  2. Deposit
  3. Withdraw
  4. Exit""")

  choice=input('enter your Choice: ')

  if choice=="1":
    print("Balance",balance)
  elif choice=="2":
    deposit_amount=int(input("Enter Deposit Amount: "))
    balance+=deposit_amount
  elif choice=="3":
    withdraw_amount=int(input("Enter Amount to withdraw: "))
    if withdraw_amount<=balance:
      print(f"You debited {withdraw_amount}.Your current balance is:{balance}")
      balance-=withdraw_amount
    else:
      print("insufficient balance")
  elif choice=="4":
    print("Thanks for banking with us")
    break
```

```python
#16.Collatz sequence: start with any number; if even, divide by 2, otherwise multiply by 3 and add 1. Repeat until you reach 1, counting the steps.

n=int(input("enter a number: "))
count=0
while n!=1:
  count+=1
  if n%2==0:
    n//=2
  else:
    n=n*3+1
  print(f"Step{count}:{n}")
print("The total steps is", count)
```

```python
#18.FizzBuzz: for 1 to 100, print “Fizz” for multiples of 3, “Buzz” for multiples of 5, “FizzBuzz” for both, otherwise the number.

for i in range(1,101):
  if i%3==0 and i%5==0:
    print("FizzBuzz",end=" ")
  elif i%3==0:
    print("Fizz", end=" ")
  elif i%5==0:
    print("Buzz", end=" ")
  else:
    print(i, end=" ")
```
