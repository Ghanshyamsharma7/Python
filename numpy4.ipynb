```python
import numpy as np
a=np.array([[2,3,4],[5,6,7]])
print(a)
```

# **Aggregation**

```python
print(a.sum())
print(a.prod())
print(a.min())
print(a.max())
print(a.mean())
```

```python
print(a.sum(axis=0))
print(a.sum(axis=1))
```

# Running **calculations**

```python
a=np.array([[2,3,4],[5,6,7]])
print(a)
print(a.cumsum())
print(a.cumsum(axis=0))
print(np.cumsum(a))
print(np.cumsum(a,axis=0))
print(np.cumsum(a,axis=1))
```

```python
print(a.cumprod())
print(np.cumprod(a))
print(np.cumprod(a,axis=0))
```

```python
print(np.diff(a))
```

```python
a=np.array([3,5,6,79,76])
print(np.diff(a))
```

# **Statistics**

```python
a=np.array([3,5,6,79,76])
print(np.mean(a))
print(np.median(a))

print(np.var(a))
print(np.std(a))
```
