```python
import numpy as np
a=np.array([[2,3,4],[5,6,7]])
print(a)
```

# **Aggregation**

```python
print(a.sum())
print(a.prod())
print(a.min())
print(a.max())
print(a.mean())
```

```python
print(a.sum(axis=0))
print(a.sum(axis=1))
```

# Running **calculations**

```python
a=np.array([[2,3,4],[5,6,7]])
print(a)
print(a.cumsum())
print(a.cumsum(axis=0))
print(np.cumsum(a))
print(np.cumsum(a,axis=0))
print(np.cumsum(a,axis=1))
```

```python
print(a.cumprod())
print(np.cumprod(a))
print(np.cumprod(a,axis=0))
```

```python
print(np.diff(a))
```

```python
a=np.array([3,5,6,79,76])
print(np.diff(a))
```

# **Statistics**

```python
a=np.array([3,5,6,79,76])
print(np.mean(a))
print(np.median(a))

print(np.var(a))
print(np.std(a))
```

```python
scores=np.array([87,88,98,89,78,92])
print(np.percentile(scores,90))
print(np.quantile(scores,0.9))
```

```python
a=np.array([22,33,44,65])
b=np.array([11,23,13,56])

print(np.corrcoef(a,b))
print(np.cov(a,b))
```

```python
a=np.array([22,33,44,65])
b=np.array([17,13,9,6])

print(np.corrcoef(a,b))
```

# **Handling NaN**

```python
a=np.array([2,3,np.nan,6,np.nan])
print(a)
print(a.sum())
print(a.prod())
```

```python
np.nan==np.nan
```

```python
np.isnan(a)
```

```python
print(np.nansum(a))
print(np.nanprod(a))
print(np.nanmean(a))
print(np.nanmax(a))
print(np.nanmin(a))
print(np.nanpercentile(a,90))
```

```python
print(a)
#drop NaN
print(a[~np.isnan(a)])
```

```python
print(np.where(np.isnan(a),0,a))
print(np.where(np.isnan(a),np.nanmean(a),a))
```
