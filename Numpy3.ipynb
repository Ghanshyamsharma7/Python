# **Vectorization**

```python
import numpy as np
a=np.array([3,2,2,5,6])
print(a)
res=a*2
print(res)
```

```python
a=np.array([3,2,2,5,6])
print(a)
res=a*2+3
print(res)
```

# **Broadcasting**

```python
arr=np.arange(12).reshape(3,4)
a=np.array([3,4,5,5])
print(arr)
print(a)
print(a.shape)
x=arr+a
print(x)
```

```python
a=np.ones(3).reshape(3,1)
print(a)
b=np.array([2,3,4])
print(a+b)
```

```python
a=np.ones(3).reshape(3,1)
print(a)
b=np.array([2,3])
print(a+b)
```

```python
a=np.ones(3)
print(a)
b=np.array([2,3])
print(b)
res=a+b[:,np.newaxis]
print(res)
```

```python
a=np.ones(3)
print(a)
b=np.array([2,3])
print(b)
res=a+b.reshape(-1,1)
print(res)
```
