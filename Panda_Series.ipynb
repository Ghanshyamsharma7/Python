```python
import numpy as np
import pandas as pd
```

# **Creating a Series**

```python
s=pd.Series([23,22,43,4,55,43])
print(s)
```

```python
s=pd.Series([23,22,43,4,55,43],index=[1,2,3,4,5,6])
print(s)
```

```python
s=pd.Series([23,22,43,4,55,43],index=[1,2,3,4,5,6],name="Numbers")
print(s)
```

```python
s=pd.Series([33,22,11],index=["a","b","c"],name="Table_name")
print(s)
print(s.index)
print(s.values)
print(s.name)
print(s.dtype)
print(s.shape)
print(s.size)
print(s.ndim)
```

```python
a=pd.Series({"Puja":89,"Ghanshyam":99,"Ragini":88})
print(a)
```

```python
a=pd.Series(5,index=["a","b","c","d"])
print(a)
```

```python
s=pd.Series(np.array([33,44,55,66,78]))
print(s)
```

# Accessing Elements

```python
#by label
s=pd.Series([33,22,11],index=["a","b","c"],name="Table_name")
print(s["a"])
print(s["a":"c"])
print(s[["a","b"]])
print(s.loc["b"])
print(s.loc[["a","b"]])
print(s.loc["a":"b"])
```

```python
# by index
print(s[0])
print(s[[0,2]])
print(s[0:3])
```

```python
#by position
print(s.iloc[0])
print(s.iloc[0:3])
print(s.iloc[[0,1]])
print(s.iloc[-1])
```

# vectorised operations

```python
s=pd.Series([33,22,11],index=["a","b","c"],name="Table_name")
print(s*5)
print(s-2)
print(s**2)
print(s%2)
print(s/3)
```

```python
# it works for all numpy functions
print(np.sum(s))
print(np.max(s))
print(np.mean(s))
```

# Filtering with Boolean Conditions

```python
marks=pd.Series([93,44,55,66,77],index=["Puja","Ghanshyam","Saloni","Geeta","Lovely"])
print(marks>60)
```

```python
print(marks[marks>50])
print(marks[(marks<90) & (marks>40)])
print(marks[(marks<50) | (marks>90)])
print(marks.isin([45,70]))
print(marks[~(marks>50)])
print(marks.between(50,80))
```
