# Accessing Elements

```python
#by label
s=pd.Series([33,22,11],index=["a","b","c"],name="Table_name")
print(s["a"])
print(s["a":"c"])
print(s[["a","b"]])
print(s.loc["b"])
print(s.loc[["a","b"]])
print(s.loc["a":"b"])
```

```python
# by index
print(s[0])
print(s[[0,2]])
print(s[0:3])
```

```python
#by position
print(s.iloc[0])
print(s.iloc[0:3])
print(s.iloc[[0,1]])
print(s.iloc[-1])
```

# Filtering with Boolean Conditions

```python
marks=pd.Series([93,44,55,66,77],index=["Puja","Ghanshyam","Saloni","Geeta","Lovely"])
print(marks>60)
```

```python
print(marks[marks>50])
print(marks[(marks<90) & (marks>40)])
print(marks[(marks<50) | (marks>90)])
print(marks.isin([45,70]))
print(marks[~(marks>50)])
print(marks.between(50,80))
```

```python
marks=pd.Series([56,67,77,87,98])
print(marks.sum())
print(marks.prod())
print(marks.mean())
print(marks.mode())
print(marks.max())
print(marks.min())
print(marks.cumsum())
print(marks.std())
print(marks.idxmax())
print(marks.idxmin())
print(marks.quantile(0.75))
print(marks.describe())
print(marks.nlargest(3))
print(marks.nsmallest(2))
```

```python
#for text or series
city=pd.Series(["Patna","Delhi","Mumbai","Kolkata","Patna"])
print(city.value_counts())
print(city.unique())
print(city.nunique())
print(city.mode())
print(city.value_counts(normalize=True))
```

```python
s=pd.Series([33,44,55],index=["a","b","c"])
s.index=["p","q","r"]
print(s["p"])
print(s)
s["q"]=88
print(s)
s=s.drop("r")
print(s)
s=s.rename("Marks")
print(s)
s=s.reset_index(drop=True)
print(s)
```

# **map apply and replace**

```python
marks=pd.Series([78,56,45,98,87])
Result=marks.apply(lambda x:"Pass" if x>50 else "Fail")
print(marks)
print(Result)
```

```python
marks.replace(56,65)
print(marks)
```

```python
marks.replace({98:89,87:78})
print(marks)
z=marks.clip(lower=50,upper=90)
print(z)
```

```python
codes=pd.Series(["M","F","F","M"])
x=codes.map({"M":"Male","F":"Female"})
print(x)
```

```python
dates=pd.Series(["2003-07-11","2000-09-12","1990-08-23"])
print(dates)
dates=pd.to_datetime(dates)
print(dates)

print(dates.dt.date)
print(dates.dt.day_name())
print(dates.dt.days_in_month)
print(dates.dt.month)
print(dates.dt.year)
```

# **String methods**

```python
name=pd.Series(["Neha","Puja","Saloni","Lovely"])

print(name.str.len())
print(name.str.capitalize())
print(name.str.lower())
# print(name.str.extract())
print(name.str.split(" "))
print(name.str.find("Puja"))
print(name.str.endswith("ly"))
print(name.str.contains("Sharma"))
print(name.str.count("a"))
print(name.str.join(" "))
```

# **Missing Data in Series**

```python
s=pd.Series([np.nan,87,np.nan,67,90,np.nan,60])

print(s.isna())
print(s.notna())
print(s.isna().sum())   #count of missing
```

```python
print(s.fillna(0))
print(s.fillna(s.mean()))
print(s.ffill())
print(s.bfill())
```
