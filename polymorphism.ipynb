```python
# Method overloading

class Calc:
  def total(self,a,b,c=0):
    return (a+b)

c=Calc()
print(c.total(3,4))
```

```python
# Method overloading

class Calc:
  def total(self,*args):
    return sum(args)

c=Calc()
print(c.total(3,4,6,7,8))
```

```python
# Method overloading (Same mthods override the previous one in python unlike java and C++)

class Calc:
  def total(self,a,b):
    return (a+b)
  def total(self,a,b,c):
    return (a+b+c)

c=Calc()
print(c.total(3,4,5))
```

```python
#Method overriding
class Animal:
  def sound(self):
    print("Animal sound")

class Dog(Animal):
  def sound(self):
    print("barks")

class Cow(Animal):
  def sound(self):
    print("maa")

c=Cow()
a=Animal()
a.sound()
c.sound()
```

```python
#Method overriding
class Animal:
  def sound(self):
    print("Animal sound")

class Dog(Animal):
  def sound(self):
    print("barks")

class Cow(Animal):
  def sound(self):
    print("meow")

animals=[Cow(),Dog(),Animal()]

for animal in animals:
  animal.sound()
```

```python
#Duck Typing

class UPI:
  def payment(self):
    print("payment done via upi")
class Debit_card:
  def payment(self):
    print("payment done via debit card")

class Credit_card:
  def payment(self):
    print("payment done via credit card")

def make_payment(pay):
  print(pay.payment())

make_payment(UPI())
make_payment(Debit_card())
make_payment(Credit_card())
```

```python
#operator overloading

class Add:
  def __init__(self,a):
    self.a=a

  def __add__(self,other):
    return (self.a+other.a)

a1=Add(46)
b2=Add(79)
print(a1+b2)
```
