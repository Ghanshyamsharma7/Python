```python
try:
  print(10/0)
except ZeroDivisionError as e:
  print(e)
```

```python
try:
  a=int(input("enter a number"))
  b=int(input("enter a number"))
  res=a/b
except ValueError :
  print("Please enter integer")

except ZeroDivisionError:
  print("Cannot divide by zero")

else:
  print(res)

finally:
  print("Program excution finished")
```
