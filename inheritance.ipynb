```python
# Single inheritance
class Animal:
  def __init__(self,name):
    self.name=name

  def info(self):
    print("Animal name:",self.name)
class Dog(Animal):
    def show(self):
      print(self.name,"Barks")

d=Dog("Alice")
d.show()
```

```python
# Multilevel inheritance
class Person:
  def __init__(self,name):
    self.name=name

class Employee(Person):
  def __init__(self,age,name):
    self.age=age
    Person.__init__(self,name)

  def show(self):
    print(f"{self.name} is {self.age} years old")

class Manager(Employee):
  def __init__(self,depart,age,name):
    self.depart=depart
    Employee.__init__(self,age,name)


  def info(self):
    print(f"{self.name} works in {self.depart} and {self.age} years old")

# m=Manager("IT")
# m.name="Puja"
# m.age=23
# m.show()
# m.info()
M=Manager("IT",32,"Ghanshyam")
M.info()
```

```python
# Multiple inheritance
class Father:
  def __init__(self,height):
    self.height=height

class Mother:
  def __init__(self,complexion):
    self.complexion=complexion


class Son(Father,Mother):
  def __init__(self,name,height,complexion):
    self.name=name
    Father.__init__(self,height)
    Mother.__init__(self,complexion)

  def show(self):
    print(f"{self.name} has acquired height {self.height} from father and complexion {self.complexion} from mother")

S=Son("Puja","tall","milky white")
S.show()
```
