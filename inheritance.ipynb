```python
# Single inheritance
class Animal:
  def __init__(self,name):
    self.name=name

  def info(self):
    print("Animal name:",self.name)
class Dog(Animal):
    def show(self):
      print(self.name,"Barks")

d=Dog("Alice")
d.show()
```

```python
# Multilevel inheritance
class Person:
  def __init__(self,name):
    self.name=name

class Employee(Person):
  def __init__(self,age,name):
    self.age=age
    Person.__init__(self,name)

  def show(self):
    print(f"{self.name} is {self.age} years old")

class Manager(Employee):
  def __init__(self,depart,age,name):
    self.depart=depart
    Employee.__init__(self,age,name)


  def info(self):
    print(f"{self.name} works in {self.depart} and {self.age} years old")

# m=Manager("IT")
# m.name="Puja"
# m.age=23
# m.show()
# m.info()
M=Manager("IT",32,"Ghanshyam")
M.info()
```

```python
# Multiple inheritance
class Father:
  def __init__(self,height):
    self.height=height

class Mother:
  def __init__(self,complexion):
    self.complexion=complexion


class Son(Father,Mother):
  def __init__(self,name,height,complexion):
    self.name=name
    Father.__init__(self,height)
    Mother.__init__(self,complexion)

  def show(self):
    print(f"{self.name} has acquired height {self.height} from father and complexion {self.complexion} from mother")

S=Son("Puja","tall","milky white")
S.show()
print(Son.__mro__)
```

```python
#Hierarchical inheritance

class shape:
  def area(self):
    pass

class Rectangle(shape):
  def __init__(self, length,breadth):
    self.length=length
    self.breadth=breadth

  def area(self):
    return self.length*self.breadth

class Circle(shape):
  def __init__(self,radius):
    self.radius=radius

  def area(self):
    return 3.14*self.radius*self.radius

rec=Rectangle(5,6)
print("Rectangle Area:",rec.area())
cir=Circle(7)
print("Circle Area:",cir.area())
```

```python
#hybrid inheritance
class University:
  def __init__(self,uni_name,**kwargs):
    self.name=uni_name
    super().__init__(**kwargs)

class Faculty(University):
  def __init__(self,sub,**kwargs):
    self.sub=sub
    super().__init__(**kwargs)

class Researcher(University):
  def __init__(self,research_topic,**kwargs):
    self.research_topic=research_topic
    super().__init__(**kwargs)

class Professor(Faculty,Researcher):
  def __init__(self,uni_name,sub,research_topic):
    super().__init__(uni_name=uni_name, sub=sub, research_topic=research_topic)

  def profile(self):
    print("University Name:",self.name,"Subject:",self.sub, "Research Topic:",self.research_topic)

P=Professor("MU","CS","AI and Humans")
P.profile()
print(Professor.__mro__)
```

```python
class A:
  def show(self):
    print("I am from A")

class B(A):
  def show(self):
    print("I m from B")

class C(A):
  def show(self):
    print("I am from C")
class D(B,C):
    pass

d=D()
d.show()

D.mro()
```
