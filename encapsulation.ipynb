```python
class Employee:
  def __init__(self,name,salary):
    self.name=name          #public attribute
    self.__salary=salary    #private attribute

E=Employee("Puja",980000)
print(E.name)
#print(E.salary)
```

```python
#protected members

class Employee:
  def __init__(self,name,department):
    self.name=name
    self._department=department            #protected attribute

class Manager(Employee):
  def show(self):
    print("Department:",self._department)  #accessible in subclass

emp=Manager("Puja","IT")
emp.show()
emp.name
```

```python
#private members

class Employee:
  def __init__(self,name,salary):
    self.name=name          #public attribute
    self.__salary=salary    #private attribute
  def show(self):
    print(f"{self.name} has {self.__salary} salary")

E=Employee("Puja",980000)
print(E.name)
E.show()            #we can access private attributes via public methods
#print(E.__salary)  #can not access private members directly
```

```python
class Employee:
  def __init__(self,name,age):
    self.name=name
    self.__age=age

  def get_age(self):
    return self.__age

  def set_age(self,age):
    if age<0:
      print("age must be positive integer")
    else:
      self.__age=age

  emp=Employee("Puja",21)
  print(emp.get_age())

  emp.set_age(-32)
  print(emp.get_age())
```

```python
class Student:
  def __init__(self,name):
    self.name=name
    self.__grade=0

  def set_grade(self,grade):
    if 0<grade<=100:
      self.__grade=grade
    else:
      print("Grade must be between 0 and 100")

  def get_grade(self):
    return self.__grade

  def check_status(self):
    if self.__grade>70:
      return "Pass"
    else:
      return "Better luck next time"

stu=Student("Puja")
stu.set_grade(89)
print(stu.get_grade())
print(stu.check_status())
```

```python
class Account:
  def __init__(self):
    self.balance=10000

  def _show_balance(self):
    print(f"Balance:{self.balance}")

  def __update_balance(self,amount):
    self.balance+=amount

  def deposit(self,amount):
    if amount>0:
      self.__update_balance(amount)
      self._show_balance()
    else:
      print("invalid amount")

acc=Account()
print(acc._show_balance())
acc.deposit(30000)
```
