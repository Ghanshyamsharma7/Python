```python
class Employee:
  def __init__(self,name,salary):
    self.name=name          #public attribute
    self.__salary=salary    #private attribute

E=Employee("Puja",980000)
print(E.name)
#print(E.salary)
```

```python
#protected members

class Employee:
  def __init__(self,name,department):
    self.name=name
    self._department=department            #protected attribute

class Manager(Employee):
  def show(self):
    print("Department:",self._department)  #accessible in subclass

emp=Manager("Puja","IT")
emp.show()
emp.name
```

```python
#private members

class Employee:
  def __init__(self,name,salary):
    self.name=name          #public attribute
    self.__salary=salary    #private attribute
  def show(self):
    print(f"{self.name} has {self.__salary} salary")

E=Employee("Puja",980000)
print(E.name)
E.show()            #we can access private attributes via public methods
#print(E.__salary)  #can not access private members directly
```
