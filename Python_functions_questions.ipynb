```python
#WAF to find the largest number in a list

def largest_in_list(li):
  largest=li[0]

  for i in range(len(li)):
    if li[i]>largest:
      largest=li[i]

  return largest

li=[34,56,34,23,67,45]
largest_in_list(li)
```

```python
#WAF that returns the number of vowels in a string
def count_vowels(str):
  count=0
  vow='aeiou'
  for ch in str:
    for v in vow:
      if v==ch:
        count+=1

  return count

count_vowels("ghanshyampuja")
```

```python
#Write a function to check whether a year is a leap year.

def if_leapYear(year):
  if (year%4==0 and year%100!=0) or (year%400==0):
    print("Leap year")
  else:
    print('not a leap year')

if_leapYear(2000)
if_leapYear(1999)
if_leapYear(2004)
```

```python
#Write a function to count the digits in a number.

def count_digit(num):
  count=0
  for i in range(len(str(num))):
    count+=1
  return count

print(count_digit(123456789))
count_digit("8877530565")
```

```python
#Write a function to calculate the product of the digits of a number.

def products_of_digits(num):
  product=1
  for digit in str(num):
    product*=int(digit)
  return product

print(products_of_digits(23445))
print(products_of_digits(45))
```

```python
#Write a function to count the even numbers in a list.

def count_even(li):
  count=0
  for i in range(len(li)):
    if li[i]%2==0:
      count+=1
  return count

li=[23,45,65,67,88,67,43,21]
print(count_even(li))
print(count_even([2,4,56,78,88,3]))
```

```python
#Write a function to find the smallest number in a list.

def smallest_in_list(li):
  min=li[0]
  for i in range(len(li)):
    if li[i]<min:
      min=li[i]
  return min

li=[33,45,76,23,99]
print(smallest_in_list(li))
```
