```python
1.#Write a function to calculate the power of a number without using **.

def power(n,ind):
  pro=1
  i=1
  while i<=ind:
    pro*=n
    i=i+1
  return pro

print(power(2,3))
print(power(3,5))
```

```python
#2.Write a function to calculate the remainder when one number is divided by another

def remainder(a,b):

  if b!=0:
    rem=a-(b*(a//b))
    return rem
  else:
    print("can not divided by zero")

print(remainder(8,3))
print(remainder(68,2))
print(remainder(68,0))
```

```python
#5.Write a function that swaps two numbers and returns the swapped values

def swap(a,b):
  temp=a
  a=b
  b=temp
  return a,b

print(swap(3,4))
```

```python
#6.Write a function to check whether a number is a perfect square.

def if_perfect_square(num):
  if num<0:
    return False

  root=int(num**0.5)
  return root*root==num

print(if_perfect_square(89))
print(if_perfect_square(81))
```

```python
#7.Write a function to check whether a number is an Armstrong number
def if_armstrong(num):
  if num<0:
    return False


  num_str=str(num)
  n=len(num_str)
  # total_sum=sum(int(digit)**n for digit in num_str)
  total=0
  for digit in num_str:
    total+=int(digit)**n
  return total==num

print(if_armstrong(153))
print(if_armstrong(123))
print(if_armstrong(1634))
```

```python
#10.Write a function to count how many numbers between 1 and n are divisible by both 3 and 5
def divisibility_of_3_5(num):
  count=0
  for i in range(1,(num+1)):
    if (i%3==0 and i%5==0):
      count+=1
  return count

divisibility_of_3_5(100)
```

```python
#11 Write a function to reverse the digits of an integer without converting it to a string.

def reverse_digits(num):
  reverse=0
  while num>0:
    digit=num%10
    reverse=(reverse*10)+digit
    num//=10
  return reverse

reverse_digits(1234)
```

```python
#12.Write a function to check whether a number is a palindrome without converting it to a string.

def is_pallindrome(num):
  original_num=num
  reverse=0
  while num>0:
    digit=num%10
    reverse=(reverse*10)+digit
    num//=10

  if reverse==original_num:
    return "Yes"
  return "No"

is_pallindrome(12321)
```

```python
#13.Write a function to find the sum of all even digits in a number.

def sum_of_eve_digits(num):
  sum=0
  while num>0:
    digit=num%10
    if digit%2==0:
      sum+=digit
    num//=10
  return sum

sum_of_eve_digits(345687)
```

```python
#14.Write a function to find the largest digit in a number.
def largest_digit(num):
  num=str(num)
  max=num[0]
  for i in range(len(num)):
    if num[i]>max:
      max=num[i]
  return max

print(largest_digit(45687))
print(largest_digit(18749))

```

```python
#16.Write a function to count uppercase and lowercase letters in a string.

def count_case(str):
  count=0
  lower_count=0
  for i in range(len(str)):
    if str[i]==str[i].upper():
      count+=1
    if str[i]==str[i].lower():
      lower_count+=1
  return count,lower_count

count,lower_count=count_case("GhaNsHYam")
print("Uppercase:",count)
print("lowercase:",lower_count)
print(count_case("PUjA"))
```
