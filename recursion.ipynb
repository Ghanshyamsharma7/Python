```python
def fact(n):
  if n==1:
    return 1
  else:
    return n*fact(n-1)
fact(5)
```

```python
def fib(n):
  if n<=1:
    return n
  else:
    return fib(n-1)+fib(n-2)
fib(7)
```

```python
def rec(n):
  if n==1:
    return n
  else:
    print(n)
    return rec(n-1)

rec(8)
print(type(rec))
```
